----------SCIKIT-LEARN

```python
#pip install scikit-learn

from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
df_scaled =df[['age','fare']].copy()
```

```python
df_scaled['age'].fillna(df_scaled['age'].median(),inplace=True)
```

```python
#initialize scalers
standard_scaler=StandardScaler()
min_max_scaler=MinMaxScaler()
```

```python
#apply StandardScaler
standard_scaler=standard_scaler.fit_transform(df_scaled)
df_standard=pd.DataFrame(standard_scaler,columns=['Age_Standard','Fare_Standard'])
```

```python
#apply MinMaxScaler
min_max_scaler=min_max_scaler.fit_transform(df_scaled)
df_min_max=pd.DataFrame(min_max_scaler,columns=['Age_MinMax','Fare_MinMax'])
```

```python
#combine orginal and scaled values
df_combined=pd.concat([df_scaled,df_standard,df_min_max],axis=1)
```

```python
#show results
print(df_combined.head())
```
